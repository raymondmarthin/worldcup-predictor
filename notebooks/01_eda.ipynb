# 01 - EDA: World Cup 2026 Knockout Predictor
Quick look at the processed datasets before feature engineering / training.

```python
import pandas as pd
import sys
from pathlib import Path
sys.path.insert(0, str(Path.cwd().parent))

matches = pd.read_csv('../data/processed/processed_matches.csv', parse_dates=['date'])
rankings = pd.read_csv('../data/processed/processed_rankings.csv', parse_dates=['date'])
wc2026 = pd.read_csv('../data/processed/processed_wc2026.csv', parse_dates=['date'])
matches.shape, rankings.shape, wc2026.shape
```

```python
# World Cup matches per edition
wc = matches[matches['tournament'] == 'FIFA World Cup']
wc['date'].dt.year.value_counts().sort_index()
```

```python
# Home win / draw / away win rate across all international matches
def result(row):
    if row.home_score > row.away_score: return 'home_win'
    if row.home_score < row.away_score: return 'away_win'
    return 'draw'

matches['result'] = matches.apply(result, axis=1)
matches['result'].value_counts(normalize=True)
```

```python
# FIFA ranking coverage over time
rankings.groupby(rankings['date'].dt.year)['team'].nunique().tail(15)
```

```python
# WC 2026 progress so far (live context, NOT used for training)
wc2026['status'].value_counts()
```
